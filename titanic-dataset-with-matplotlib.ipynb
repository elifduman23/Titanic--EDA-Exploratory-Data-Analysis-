```python
import pandas as pd #pandas kütüphanesini etkinleştirdik.
import matplotlib.pyplot as plt #matplotlib kütüphanesini etkinleştirdik.
data=pd.read_csv("/kaggle/input/datasets/elifduman/titanic-dataset/titanic.csv")
```

```python
age=pd.DataFrame(data["Age"]) #age kolonundaki verileri DataFrame yapıp age değişkenine atadık.
print(age)
```

```python
name =pd.DataFrame(data["Name"])
print(name)
```

```python
print(data.shape) # verinin kaç satır olduğuna bakarız.
```

```python
plt.hist(age)#yolcuların yaş dağılımına histogram grafiği ile bakarız
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(data["Age"].dropna(),bins=20,color="red",edgecolor="black")
plt.title("Yolcuların yaş dağılımı")
plt.xlabel("Yaş")
plt.ylabel("Kişi sayısı")
plt.show()
```

```python
Sex_counts=data["Sex"].value_counts()
plt.figure(figsize=(6,4))
plt.bar(Sex_counts.index,Sex_counts.values,color=["blue","pink"])
plt.title("Cinsiyet Dağılımı")
plt.ylabel("Kişi Sayısı")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(data["Age"],data["Fare"],alpha=0.5,color="Purple")
plt.title("Yaş ve bilet fiyatları")
plt.xlabel("Yaş")
plt.ylabel("Bilet fiyatı")
plt.show()
```

```python
Embarkeds=data["Embarked"].value_counts()
plt.figure(figsize=(8,4))
plt.barh(Embarkeds.index,Embarkeds.values,color=["lightgreen","orange","grey"])
plt.title("Hangi limandan kaç kişi bindi")
plt.xlabel("Yolcu Sayısı")
plt.ylabel("Liman kodu")
plt.show()
```
